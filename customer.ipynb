```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
df = df.rename(columns={"purchase_amount_(usd)": "purchase_amount"})
```

```python
print(df.columns)
```

```python
#crear la columna age_group
labels = ["niño","joven","adulto","anciano"]
df["age_group"] = pd.qcut(df["age"],q=4,labels=labels)
```

```python
print(df.columns)
```

```python
labels = ["Young_adult","Adult","Middle_aged","Senior"]
df["age_group"] = pd.qcut(df["age"],q=4,labels=labels)
```

```python
df.age_group
```

```python
df[['age','age_group']].head(10)
```

```python
# crear columna de frecuencia compras por dia o dias
frequency_mapping = {
    "Fortnightly": 14,
    "Weekly":7,
    "Monthly":(28,30,31),
    "Quarterly":90,
    "Bi-Weekly":14,
    "Annually":365,
    "Every 3 Months":90
}
df["Purchases_Frecuentry_Days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["frequency_of_purchases","Purchases_Frecuentry_Days"]].head(10)
```
